```python
# CELLULE 9 - LOAD: Insert data into MySQL
# But: Test inserting DataFrames into MySQL tables

from sqlalchemy import create_engine, text
from dotenv import load_dotenv
import os
from urllib.parse import quote_plus
import time

# 1. Load connection info
load_dotenv()

DB_USER = 'root'
DB_PASSWORD = os.getenv('MYSQL_ROOT_PASSWORD')
DB_HOST = os.getenv('MYSQL_HOST', 'localhost')
DB_PORT = os.getenv('MYSQL_PORT', '2004')
DB_NAME = os.getenv('MYSQL_DATABASE', 'movie_recommendation')

# 2. Create connection
encoded_password = quote_plus(DB_PASSWORD)
connection_string = f"mysql+mysqlconnector://{DB_USER}:{encoded_password}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
engine = create_engine(connection_string)

print("="*50)
print("📤 LOAD TEST - Inserting data into MySQL")
print("="*50)

# 3. Check current state of tables
with engine.connect() as conn:
    movie_count_before = conn.execute(text("SELECT COUNT(*) FROM movies")).scalar()
    rating_count_before = conn.execute(text("SELECT COUNT(*) FROM ratings")).scalar()

print(f"\n📊 BEFORE INSERT:")
print(f"   Movies in table: {movie_count_before}")
print(f"   Ratings in table: {rating_count_before}")

# 4. Insert movies
print("\n🎬 Inserting movies...")
start = time.time()
df_movies.to_sql('movies', engine, if_exists='append', index=False)
end = time.time()
print(f"   ✅ {len(df_movies)} movies inserted in {end-start:.2f} seconds")

# 5. Insert ratings
print("\n⭐ Inserting ratings...")
start = time.time()
df_ratings.to_sql('ratings', engine, if_exists='append', index=False)
end = time.time()
print(f"   ✅ {len(df_ratings)} ratings inserted in {end-start:.2f} seconds")

# 6. Verify final state
with engine.connect() as conn:
    movie_count_after = conn.execute(text("SELECT COUNT(*) FROM movies")).scalar()
    rating_count_after = conn.execute(text("SELECT COUNT(*) FROM ratings")).scalar()

print(f"\n📊 AFTER INSERT:")
print(f"   Movies in table: {movie_count_after}")
print(f"   Ratings in table: {rating_count_after}")

# 7. Show sample from database
print("\n👀 Sample from movies table:")
sample_movies = pd.read_sql("SELECT * FROM movies LIMIT 5", engine)
print(sample_movies)

print("\n👀 Sample from ratings table:")
sample_ratings = pd.read_sql("SELECT * FROM ratings LIMIT 5", engine)
print(sample_ratings)

print("\n" + "="*50)
print("✅ LOAD TEST COMPLETED")
print("="*50)
```

```python
# CELLULE 11 - LOAD with REPLACE
# But: Replace all data in both tables (clean start)

from sqlalchemy import create_engine, text
from dotenv import load_dotenv
import os
from urllib.parse import quote_plus
import time
import pandas as pd

load_dotenv()

DB_USER = 'root'
DB_PASSWORD = os.getenv('MYSQL_ROOT_PASSWORD')
DB_HOST = os.getenv('MYSQL_HOST', 'localhost')
DB_PORT = os.getenv('MYSQL_PORT', '2004')
DB_NAME = os.getenv('MYSQL_DATABASE', 'movie_recommendation')

encoded_password = quote_plus(DB_PASSWORD)
connection_string = f"mysql+mysqlconnector://{DB_USER}:{encoded_password}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
engine = create_engine(connection_string)

print("="*50)
print("📤 LOAD TEST - Replacing all data")
print("="*50)

# Check before
with engine.connect() as conn:
    movie_before = conn.execute(text("SELECT COUNT(*) FROM movies")).scalar()
    rating_before = conn.execute(text("SELECT COUNT(*) FROM ratings")).scalar()

print(f"\n📊 BEFORE REPLACE:")
print(f"   Movies: {movie_before}")
print(f"   Ratings: {rating_before}")

# Insert movies (replace)
print("\n🎬 Replacing movies...")
start = time.time()
df_movies.to_sql('movies', engine, if_exists='replace', index=False)
end = time.time()
print(f"   ✅ {len(df_movies)} movies inserted in {end-start:.2f} seconds")

# Insert ratings (replace)
print("\n⭐ Replacing ratings...")
start = time.time()
df_ratings.to_sql('ratings', engine, if_exists='replace', index=False)
end = time.time()
print(f"   ✅ {len(df_ratings)} ratings inserted in {end-start:.2f} seconds")

# Check after
with engine.connect() as conn:
    movie_after = conn.execute(text("SELECT COUNT(*) FROM movies")).scalar()
    rating_after = conn.execute(text("SELECT COUNT(*) FROM ratings")).scalar()

print(f"\n📊 AFTER REPLACE:")
print(f"   Movies: {movie_after}")
print(f"   Ratings: {rating_after}")

# Show samples
print("\n👀 Sample from movies:")
sample_movies = pd.read_sql("SELECT * FROM movies LIMIT 3", engine)
print(sample_movies)

print("\n👀 Sample from ratings:")
sample_ratings = pd.read_sql("SELECT * FROM ratings LIMIT 3", engine)
print(sample_ratings[['userId', 'movieId', 'rating', 'timestamp']])

print("\n" + "="*50)
print("✅ LOAD TEST COMPLETED")
print("="*50)
```

```python
# CELLULE 11 - LOAD with REPLACE (Correct order)
# But: Replace all data respecting foreign key constraints

from sqlalchemy import create_engine, text
from dotenv import load_dotenv
import os
from urllib.parse import quote_plus
import time
import pandas as pd

load_dotenv()

DB_USER = 'root'
DB_PASSWORD = os.getenv('MYSQL_ROOT_PASSWORD')
DB_HOST = os.getenv('MYSQL_HOST', 'localhost')
DB_PORT = os.getenv('MYSQL_PORT', '2004')
DB_NAME = os.getenv('MYSQL_DATABASE', 'movie_recommendation')

encoded_password = quote_plus(DB_PASSWORD)
connection_string = f"mysql+mysqlconnector://{DB_USER}:{encoded_password}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
engine = create_engine(connection_string)

print("="*50)
print("📤 LOAD TEST - Replacing all data (correct order)")
print("="*50)

# Check before
with engine.connect() as conn:
    movie_before = conn.execute(text("SELECT COUNT(*) FROM movies")).scalar()
    rating_before = conn.execute(text("SELECT COUNT(*) FROM ratings")).scalar()

print(f"\n📊 BEFORE REPLACE:")
print(f"   Movies: {movie_before}")
print(f"   Ratings: {rating_before}")

# Step 1: Drop tables in correct order (child first, then parent)
print("\n🗑️  Dropping tables in correct order...")
with engine.connect() as conn:
    # Disable foreign key checks temporarily (for MySQL)
    conn.execute(text("SET FOREIGN_KEY_CHECKS = 0"))
    conn.execute(text("DROP TABLE IF EXISTS ratings"))
    conn.execute(text("DROP TABLE IF EXISTS movies"))
    conn.execute(text("SET FOREIGN_KEY_CHECKS = 1"))
    conn.commit()
print("   ✅ Tables dropped")

# Step 2: Insert movies (this will recreate the table)
print("\n🎬 Creating and inserting movies...")
start = time.time()
df_movies.to_sql('movies', engine, if_exists='replace', index=False)
end = time.time()
print(f"   ✅ {len(df_movies)} movies inserted in {end-start:.2f} seconds")

# Step 3: Insert ratings (this will recreate the table)
print("\n⭐ Creating and inserting ratings...")
start = time.time()
df_ratings.to_sql('ratings', engine, if_exists='replace', index=False)
end = time.time()
print(f"   ✅ {len(df_ratings)} ratings inserted in {end-start:.2f} seconds")

# Check after
with engine.connect() as conn:
    movie_after = conn.execute(text("SELECT COUNT(*) FROM movies")).scalar()
    rating_after = conn.execute(text("SELECT COUNT(*) FROM ratings")).scalar()

print(f"\n📊 AFTER REPLACE:")
print(f"   Movies: {movie_after}")
print(f"   Ratings: {rating_after}")

# Show samples
print("\n👀 Sample from movies:")
sample_movies = pd.read_sql("SELECT * FROM movies LIMIT 3", engine)
print(sample_movies)

print("\n👀 Sample from ratings:")
sample_ratings = pd.read_sql("SELECT * FROM ratings LIMIT 3", engine)
print(sample_ratings[['userId', 'movieId', 'rating']])

print("\n" + "="*50)
print("✅ LOAD TEST COMPLETED")
print("="*50)
```

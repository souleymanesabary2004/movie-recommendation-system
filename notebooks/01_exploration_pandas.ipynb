```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print(" libraries imported:")
print(f"Pandas version: {pd.__version__}")
print(f"NumPy version: {np.__version__}")
print(f"Matplotlib version: {plt.matplotlib.__version__}")
```

```python
movies_path='../data/raw/movies.csv'
ratings_path='../data/raw/ratings.csv'
print(movies_path)
print('files loading (Extract)')
df_movies=pd.read_csv(movies_path)
df_ratings=pd.read_csv(ratings_path)
print(f"\n Films loaded:{len(df_movies)} rows")
print(f"\n Ratings loaded:{len(df_ratings)} rows")
print("\n"+"="*50)
print("Movies review:")
print("="*50)
display(df_movies.head(3))
print("\n"+"="*50)
print("Ratings review")
print("="*50)
display(df_ratings.head(3))
```

```python
print("\n"+"="*50)
print("MOVIES DATASET INFO:")
print("="*50)
print("\nShape:", df_movies.shape)
print("\nColumns:", df_movies.columns.tolist())
print("\nData Types:")
print(df_movies.dtypes)
print("\Missing Values:")
print(df_movies.isnull().sum())

print("\n"+"="*50)
print("RATINGS DATASET INFO:")
print("="*50)
print("\nShapes:", df_ratings.shape)
print("\nColumns",df_ratings.columns.tolist())
print("\nData Types:")
print(df_ratings.dtypes)
print("\nMissing Values:")
print(df_ratings.isnull().sum())
```

```python
# Statistiques sur les notes
print("="*50)
print("RATINGS STATISTICS:")
print("="*50)
display(df_ratings['rating'].describe())

# Distribution des notes (en chiffres)
print("\n" + "="*50)
print("RATINGS DISTRIBUTION:")
print("="*50)
print(df_ratings['rating'].value_counts().sort_index())

# Distribution des genres (top 10)
print("\n" + "="*50)
print("TOP 10 GENRES:")
print("="*50)
# Les genres sont séparés par |, on va les compter
all_genres = df_movies['genres'].str.split('|', expand=True).stack()
print(all_genres.value_counts().head(10))
```

```python
  # Histogramme Of rates Pro version

sns.set_style('whitegrid')
plt.figure(figsize=(10, 4.5))  # thinks to matplotlib 

  # plt.hist(df_ratings['rating'], bins=19, color='skyblue', edgecolor='black',alpha=0.8)  # 100% version of matplotlib 
 # sns.histplot(df_ratings['rating'], kde=True)     # 100% seaborn version

sns.histplot(df_ratings['rating'],
             kde=True,
             color='steelblue',
             edgecolor='black',
             alpha=0.8)
plt.title('Distribution of Ratings', fontsize=16)
plt.xlabel('Rating', fontsize=12)
plt.ylabel('Number of Ratings', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```
